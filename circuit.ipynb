```python
import numpy as np
import matplotlib.pyplot as plt


R = 1.644e3      
L = 3.32e-9      
C = 1.886e-12    


f_GHz = np.linspace(0.5, 3.0, 2000)
f = f_GHz * 1e9  
w = 2 * np.pi * f

w0 = 1 / np.sqrt(L * C)
A = (w * L) / R
B = 1 - (w / w0)**2

R_rlc = (w * L * A) / (A**2 + B**2)
X_rlc = (w * L * B) / (A**2 + B**2)

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 8))

ax1.plot(f_GHz, R_rlc, 'k-' )
ax1.set_yscale('log')
ax1.set_xlim(0.5, 3.0)
ax1.set_ylim(0.01, 10000)
ax1.set_xlabel('Frequency (GHz)')
ax1.set_ylabel('Resistance (Ohms)')
ax1.set_title('Surface Resistance vs. Frequency')
ax1.grid(True, linestyle=':')



ax2.plot(f_GHz, X_rlc, 'k-')
ax2.set_xlim(0.5, 3.0)
ax2.set_ylim(-1000, 1000)
ax2.set_xlabel('Frequency (GHz)')
ax2.set_ylabel('Reactance (Ohms)')
ax2.set_title('Surface Reactance vs. Frequency')
ax2.grid(True, linestyle=':')


plt.tight_layout()
plt.show()
```
